```python
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
df = pd.read_csv('results.csv')
```

```python

x = df['num_vms']
plt.title('Servers Required')
plt.plot(x, df['Cosine'], label='Cosine-Best-Fit')
plt.plot(x, df['ff'], label='Fit-first')
plt.plot(x, df['Harmonic'], label='Harmonic-Fit-first')
plt.plot(x, df['Clustering'], label='Cluster-Fit-first')
plt.xlabel('Number of VMs')
plt.ylabel('Number of Servers Required')
plt.legend()
```
